### Existing Fatigue Measurement Methods

1.   **Subjective methods** - Questionaire,Self reporting
2.   **Performance methods** - Reaction tests,task performance
3.   **Biomathematical models** - Prediction models using sleep work-shedule
4.   **Behaviour-based methods** - Observing face,eyes using computer vision
5.   **Physiological Signal methods** - ECG,EEG,EMG,PPG,skin temperature



Even though physiological signals are powerful, they have some problems:

1. **Large differences between individuals**

* Some people show large changes in heart rate when tired
* Others show small changes
* Some people’s EEG patterns vary more than others

This makes it hard to set one universal threshold for detecting fatigue.

2. **Physiology is influenced by many other factors**

Fatigue is not the only thing that changes these signals:

* emotions (stress, anxiety)
* environmental temperature
* illness or medical conditions
* noise or vibration
* physical activity levels

So the system may misinterpret these changes as fatigue.

3. **Real-life reliability is still uncertain**

Most accurate physiological research is done in controlled lab environments.
But in real-life:

* people move
* sensors shift
* data gets noisy
* environment is unpredictable

Therefore, the accuracy of physiological fatigue detection in daily life still needs improvement.
