**Fatigue results from many systems: muscles, metabolism, cardiovascular system, thermoregulation, brain.**

**Why Is Fatigue Hard to Measure?**

Fatigue is complex because it involves:

- Local muscle fatigue
- Central factors (metabolic, cardiovascular, temperature regulation)

There is no single standard method to measure fatigue.
<br><br>

### Existing Fatigue Measurement Methods

1.   **Subjective methods** - Questionaire,Self reporting
2.   **Performance methods** - Reaction tests,task performance
3.   **Biomathematical models** - Prediction models using sleep work-shedule
4.   **Behaviour-based methods** - Observing face,eyes using computer vision
5.   **Physiological Signal methods** - ECG,EEG,EMG,PPG,skin temperature



Even though physiological signals are powerful, they have some problems:

1. **Large differences between individuals**

* Some people show large changes in heart rate when tired
* Others show small changes
* Some people’s EEG patterns vary more than others

This makes it hard to set one universal threshold for detecting fatigue.

2. **Physiology is influenced by many other factors**

Fatigue is not the only thing that changes these signals:

* emotions (stress, anxiety)
* environmental temperature
* illness or medical conditions
* noise or vibration
* physical activity levels

So the system may misinterpret these changes as fatigue.

3. **Real-life reliability is still uncertain**

Most accurate physiological research is done in controlled lab environments.
But in real-life:

* people move
* sensors shift
* data gets noisy
* environment is unpredictable

Therefore, the accuracy of physiological fatigue detection in daily life still needs improvement.
<br><br>

**Aerobic Fatigue Threshold**  

The anaerobic threshold (AT), also called the lactate threshold, is the point during hard exercise when your body starts making lactate as fast as at the same rate that it can remove that lactate from the blood. Before this point, your muscles can keep working without much trouble. After you pass it, lactate builds up quickly in the blood, making your muscles feel tired, your breathing faster, and it harder for oxygen to reach your muscles, which causes fatigue and reduces performance. The anaerobic threshold is important because it shows how good your endurance and fitness are, and training near this level can help you exercise longer before getting tired.

#### AFT = Avg oxygen consumption (VO₂) / Max oxygen consumption (VO₂max)

- But for this we need VO₂max (MAC) for each worker and each construction activity.
- But measuring VO₂max directly is difficult and expensive.
- So this objective creates models to **estimate MAC using submaximal exercise methods**, which are cheaper and easier.

**What is thermoregulation?**

Your body controls temperature by:

- **Vasoconstriction** → blood vessels shrink, less blood goes to skin (cold)
- **Vasodilation** → blood vessels widen, more blood goes to skin (heat)

These changes can be measured using thermal sensors.g

**Why measure skin temperature instead of blood flow?**

Direct skin blood flow measurement methods (e.g., plethysmography, Doppler, isotopes) are:

- Very sensitive to movement
- Require exact placement
- Sometimes need injections
- Not suitable for construction workers

So researchers use infrared skin temperature sensors(MLX90614) as an alternative because:

- Non-contact
- Fast
- Reliable
- Work well during movement

**Skin temperature (thermoregulation) predicts fatigue BETTER than heart rate.Temple sensors give the cleanest temperature signals = better fatigue prediction**

Below found from this research Paper - https://www.sciencedirect.com/science/article/pii/S0925753523001844

Most studies used Borg-20 RPE as a surrogate for fatigue.However, Borg-20 measures exertion, not fatigue.

Studies predicting exertion assume exertion ≈ fatigue.Exertion is the immediate effort level.Fatigue is a long-term decline in capacity.

- **Fatigue** - 11-point scale (0 to 10), where 0 is minimal fatigue and 10 is maximal exhaustion. This was the study's primary measure of perceived fatigue.(Scale used - Ratings of Fatigue (ROF) Scale)

- **Exertion** - Scale of 6 (no exertion) to 20 (maximal exertion). This measures the subjective perception of the intensity of the physical activity.(Scale used -Borg-20 Scale (RPE))

**Verification: Proving Exertion is not same as Fatigue**

- **How They Did It**: They asked the 14 workers for two scores at the same time: the Exertion (RPE) score and the Fatigue (ROF) score.

- **The Test**: They used a statistical test ($\text{Kendall's } \tau_b$) to see if these two scores matched up perfectly for every person.

- **The Finding**: For some workers, the scores matched (high correlation), but for others, they did not match at all (correlation near $0.05$).

- **Conclusion**: Exertion and Fatigue are not reliably the same. You cannot use the easy "How hard does it feel?" score to accurately guess the "How tired are you?" score for everyone.

**Prediction: Finding a Non-Invasive Substitute**

Now that they knew the Exertion score was a bad predictor, they needed a non-invasive way to predict the Fatigue score.

- **The Tool**: The wearable sensor data (EDA, HR, HRV) is the non-invasive tool.

Sensor used :

- PPG (Photoplethysmography)
- Accelerometer/Gyroscope
- EDA/GSR (Electrodermal Activity/Galvanic Skin Response)
- Skin Temperature

- **The Question**: "Do the continuous sensor signals change reliably when the worker reports a higher Fatigue (ROF) score?"

- **How They Did It**: They took the sensor data (the average signal for every 30 seconds) and compared it statistically to the reported Fatigue (ROF) score.

- **The Finding**: The sensor data did change reliably with the Fatigue score. **The EDA sensor (skin activity) was the best predictor**, reliably telling the difference between low, medium, and high fatigue levels.
