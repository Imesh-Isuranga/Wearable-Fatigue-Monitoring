### Why Workers Get WMSDs(Work-Related Musculoskeletal Disorders)

**Physical & job-related causes**

* Repetitive motions
* Awkward postures
* Heavy physical exertion
* Vibration exposure
* Ignoring safety protocols
* Severe temperatures (hot/cold)

**Environmental causes**

* Heat stress
* Cold stress
* Rapid temperature changes
<br>

**Why Is Fatigue Hard to Measure?**

Fatigue is complex because it involves:

- Local muscle fatigue
- Central factors (metabolic, cardiovascular, temperature regulation)

There is no single standard method to measure fatigue.
<br><br>

### Existing Fatigue Measurement Methods

1.   **Subjective methods** - Questionaire,Self reporting
2.   **Performance methods** - Reaction tests,task performance
3.   **Biomathematical models** - Prediction models using sleep work-shedule
4.   **Behaviour-based methods** - Observing face,eyes using computer vision
5.   **Physiological Signal methods** - ECG,EEG,EMG,PPG,skin temperature



Even though physiological signals are powerful, they have some problems:

1. **Large differences between individuals**

* Some people show large changes in heart rate when tired
* Others show small changes
* Some people’s EEG patterns vary more than others

This makes it hard to set one universal threshold for detecting fatigue.

2. **Physiology is influenced by many other factors**

Fatigue is not the only thing that changes these signals:

* emotions (stress, anxiety)
* environmental temperature
* illness or medical conditions
* noise or vibration
* physical activity levels

So the system may misinterpret these changes as fatigue.

3. **Real-life reliability is still uncertain**

Most accurate physiological research is done in controlled lab environments.
But in real-life:

* people move
* sensors shift
* data gets noisy
* environment is unpredictable

Therefore, the accuracy of physiological fatigue detection in daily life still needs improvement.
<br><br>

**Problems with current fatigue monitoring methods:**

* Subjective
* Not real-time
* Cannot prevent accidents
* Fail under extreme temperatures
<br>

### Important observations:

* **EEG is the strongest signal for brain-related fatigue (drowsiness, vigilance).**


* **PPG and ECG** are very useful for heart-based fatigue factors (HR, HRV).

* **IMU** strongly contributes to physical and muscle fatigue detection.

* **Eye movements**/blinks are powerful for drowsiness detection.

* **Temperature** helps in both mental and physical fatigue studies.

<br>To handle large differences between subjects, researchers used:

* **Normalization methods** to reduce inter-person variability

* **Demographic data** (age, height, weight)

* **Contextual data**(sleep quality, circadian rhythm, working experience)

Adding these improved model accuracy significantly.
<br><br>
**Wearables + IoT + BIM allow:**

* Continuous health tracking
* Real-time data transmission
* Visualizing worker health on site maps
* Predicting unsafe conditions before injury occurs
<br><br>

### Current Research Gaps

**Existing research limitations:**

1. Many studies done only in the lab, not real construction sites.
2. Most do NOT include **temperature effects.**
3. Physiological + environmental + task exertion combined models are rare.
4. Accuracy drops under:
* extreme heat
* extreme cold
* mixed environmental conditions
5. Sensor data often has:
* noise
* placement errors
* variations between workers
<br>

**Types of fatigue indicators:**

* Muscle fatigue (EMG)
* Cardiovascular fatigue (HR, HRV)
* Perceived fatigue (ROF)

<br><br>
Fatigue depends on which activity the worker is doing.
So the system must automatically identify activities like:

- lifting
- pushing
- pulling
- assembly tasks
- repetitive movements
- whole-body motions
<br>

Activity recognition enables:
- Automatic AFT calculation
- Continuous fatigue monitoring
- Worker productivity and safety assessment
<br>

**Aerobic Fatigue Threshold**  
#### AFT = Avg oxygen consumption (VO₂) / Max oxygen consumption (VO₂max)

- But for this we need VO₂max (MAC) for each worker and each construction activity.
- But measuring VO₂max directly is difficult and expensive.
- So this objective creates models to **estimate MAC using submaximal exercise methods**, which are cheaper and easier.
